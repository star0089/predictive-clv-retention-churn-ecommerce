## What the data can and cannot answer

Measured from Olist:
- delivered orders and customer-paid order value;
- RFM segments;
- monthly purchase activity retention;
- predicted 240-day no-return risk;
- predicted 90-, 180- and 365-day customer-paid value, when models train.

Not measured by the standard Olist data:
- contribution margin and profit CLV;
- CAC and paid marketing ROAS;
- customer-support cost and ticket causes;
- full refund/return cost;
- campaign treatment uplift.

Do not fill these gaps with invented values. If real external tables are later obtained, document their provenance and join keys before calculating these metrics.


## Decision-support concept

A paid-value-at-risk proxy can be calculated as churn probability multiplied by predicted 365-day paid value. It helps rank customer exposure, but it is not expected saved value. A customer can have high churn risk and still be impossible to retain. Treatment selection needs controlled experiments or uplift modeling.
